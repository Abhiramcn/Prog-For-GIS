```python
def count_vowels1(text):
    count = 0
    vowels = "aeiouAEIOU"

    # Loop through each character in the string
    for char in text:
         if char in vowels:
            count += 1

    return count

# Example usage:
print(count_vowels1(input("Enter the string")))
```

```python
def count_vowels2(input_string):
    # Initialize dictionary with vowels set to 0
    vowel_counts = {'a': 0, 'e': 0, 'i': 0, 'o': 0, 'u': 0}

    # Iterate through each character in the string
    for char in input_string.lower():
        if char in vowel_counts:
            vowel_counts[char] += 1

    return vowel_counts

# Example Usage:
text = input("Enter the string")
print(count_vowels2(text))

```

```python
class operations:
    def __init__(self, number):
        self.number = number

    def Addition(self, other):
        return self.number + other

    def subtraction(self, other):
        return self.number - other

    def multiplication(self, other):
        return self.number * other

    def division(self, other):
        return self.number / other


# Example usage
calculation = operations(20)
print(calculation.Addition(5))
print(calculation.subtraction(5))
print(calculation.multiplication(5))
print(calculation.division(5))
```

```python
def odde10():
    numbers = list(map(int, input("Enter numbers separated by spaces: ").split()))

    even_numbers = [number for number in numbers if number % 2 == 0]
    odd_numbers = [number for number in numbers if number % 2 != 0]

    print("Even numbers:", even_numbers)
    print("Odd numbers:", odd_numbers)
odde10()    
```

```python
def maxProfit(price_by_date):
    dates = sorted(price_by_date)
    lowest_price = price_by_date[dates[0]]
    buy_date = dates[0]
    maximum_profit = 0
    best_buy_date = dates[0]
    best_sell_date = dates[0]

    for date in dates[1:]:
        price = price_by_date[date]

        if price - lowest_price > maximum_profit:
            maximum_profit = price - lowest_price
            best_buy_date = buy_date
            best_sell_date = date

        if price < lowest_price:
            lowest_price = price
            buy_date = date

    return maximum_profit, best_buy_date, best_sell_date


price_by_date = {
    "2026-09-01": 7,
    "2026-09-02": 1,
    "2026-09-03": 5,
    "2026-09-04": 3,
    "2026-09-05": 6,
    "2026-09-06": 4
}

profit, buy_date, sell_date = maxProfit(price_by_date)
print("Maximum profit:", profit)
print("Buy on:", buy_date)
print("Sell on:", sell_date)
```

```python
def longestConsecutive(nums):
    numbers = sorted(nums)

    if not numbers:
        return 0

    longest_sequence = 1
    current_sequence = 1

    for index in range(1, len(numbers)):
        if numbers[index] == numbers[index - 1] + 1:
            current_sequence += 1
        else:
            current_sequence = 1

        longest_sequence = max(longest_sequence, current_sequence)

    return longest_sequence


nums = [100, 4, 200, 1, 3, 2]
print("Length of longest consecutive sequence:", longestConsecutive(nums))
```
