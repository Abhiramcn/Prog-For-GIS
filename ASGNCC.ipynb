```python
import time
import matplotlib.pyplot as plt
```

```python
def sum_of_cubes(N):
    total = 0

    for i in range(1, N + 1):
        total += i**3

    return total
```

```python
N_values = [100, 500, 1000, 2000, 5000, 10000, 20000, 50000]
running_times = []

for N in N_values:
    start = time.perf_counter()
    result = sum_of_cubes(N)
    elapsed_time = time.perf_counter() - start
    running_times.append(elapsed_time)

    print("N =", N)
    print("Sum =", result)
    print("Running time =", elapsed_time, "seconds")
    print("------------------------")
```

```python
plt.plot(N_values, running_times, marker='o')
plt.xlabel("N")
plt.ylabel("Running Time (seconds)")
plt.title("Running Time vs N for Sum of Cubes")
plt.grid(True)
plt.show()
```
